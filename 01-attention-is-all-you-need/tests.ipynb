## Decoder Block + Full flow test

```python
import torch.nn as nn 

from Layers.Embeddings import EmbeddingBlock
from Layers.PositionalEncoding import PositionalEncoding

from Encoder import EncoderBlock
from Decoder import DecoderBlock

max_src_len = 7
max_tgt_len = 9

input_tokens = torch.Tensor([[382, 747, 472, 472, 72]]).to(torch.long)
input_tokens = nn.functional.pad(input_tokens, (0,max_src_len - input_tokens.shape[1]), mode='constant', value=1)

src_masking = input_tokens != 1
## Mask shape must be (b_size, heads, seq_len, seq_len) to be able to broadcast so (b_size, 1, 1, seq_len)
src_masking = src_masking.unsqueeze(1).unsqueeze(1)
# Encoder
enc_embedding_layer = EmbeddingBlock(20000, D_MODEL)
enc_positional_encoding = PositionalEncoding(max_src_len, D_MODEL, 0.2)
## Pre Encoder part
enc_embeddings = enc_embedding_layer(input_tokens)
enc_pos_encoding_output = enc_positional_encoding(enc_embeddings)
## Encoder 
encoder = EncoderBlock(D_MODEL, H)
encoder_output = encoder(enc_pos_encoding_output, src_masking)


# Decoder
output_tokens = torch.Tensor([[231, 243, 981, 276, 32]]).to(torch.long)
output_tokens = nn.functional.pad(output_tokens, (0,max_tgt_len - output_tokens.shape[1]), mode='constant', value=1)
tgt_masking = output_tokens != 1
tgt_masking = tgt_masking.unsqueeze(1).unsqueeze(1)

causal_masking = torch.tril(torch.ones(max_tgt_len, max_tgt_len), diagonal=0).bool()
tgt_masking = tgt_masking & causal_masking

dec_embedding_layer = EmbeddingBlock(20000, D_MODEL)
dec_positional_encoding = PositionalEncoding(max_tgt_len, D_MODEL, 0.2)

dec_embeddings = dec_embedding_layer(output_tokens)
dec_pos_encoding_output = dec_positional_encoding(dec_embeddings)

decoder = DecoderBlock(D_MODEL, H)
decoder_output = decoder(dec_pos_encoding_output, encoder_output, src_masking, tgt_masking)
```
